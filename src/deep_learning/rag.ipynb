```python
import os
from dotenv import load_dotenv
load_dotenv()

os.getenv("LANGCHAIN_API_KEY")
os.getenv("LANGCHAIN_TRACING_V2")
os.getenv("LANGSMITH_PROJECT")


```

```python
from langchain_community.document_loaders import WebBaseLoader

loader=WebBaseLoader(web_path="https://register.educosys.com/new-courses/29-genai")

docs=loader.load()
print(docs)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits=text_splitter.split_documents(docs)
print(splits)
```

```python
print("split-0:",splits[0])
print("split-1:",splits[1])
print("split-2:",splits[2])
print("split-3:",splits[3])
```

```python
from sentence_transformers import SentenceTransformer
# Free, runs locally
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

def get_local_embedding(text):
    return embedding_model.encode(text).tolist()

# I passed a raw embedding vector instead of an Embeddings interface object,
# which LangChain requires to dynamically embed documents and queries.
```
